```python
import numpy as np

class Loss_CategoricalCrossentropy:
    # Forward pass to calculate the categorical cross-entropy loss
    def forward(self, y_pred, y_true):
        # Number of samples in the batch
        samples = len(y_pred)
        
        # Clip data to prevent division by zero (and log(0) error)
        # Clip both ends to avoid very high or low values, which could cause instability
        y_pred_clipped = np.clip(y_pred, 1e-7, 1 - 1e-7)
        
        # Check if labels are one-hot encoded or integer class labels
        if len(y_true.shape) == 1:
            # If y_true contains integer labels, pick the predicted probability for the correct class
            correct_confidences = y_pred_clipped[range(samples), y_true]
            
        elif len(y_true.shape) == 2:
            # If y_true is one-hot encoded, multiply the predicted probabilities by the correct class labels
            correct_confidences = np.sum(y_pred_clipped * y_true, axis=1)
            # print(correct_confidences)
        
        # Calculate the negative log-likelihood for each sample
        negative_log_likelihoods = -np.log(correct_confidences)
        
        return negative_log_likelihoods
    
    # Calculate method to compute the mean loss over a batch
    def calculate(self, y_pred, y_true):
        sample_losses = self.forward(y_pred, y_true)
        data_loss = np.mean(sample_losses)
        return data_loss

# Example usage:

# Assume this is the output from a softmax layer (batch of 3 samples, 3 classes)
y_pred = np.array([[0.7, 0.2, 0.1],
                   [0.1, 0.5, 0.4],
                   [0.02, 0.9, 0.08]])

# Assume these are the true labels (either as class indices or one-hot encoded)
# y_true = np.array([0, 1, 1])  # Class indices for each sample
y_true = np.array([[1, 0, 0],
                    [0, 1, 0],
                    [0, 1, 0]])

# Create a loss instance and calculate the loss
loss_function = Loss_CategoricalCrossentropy()
loss = loss_function.calculate(y_pred, y_true)

print("Loss:", loss)
```
