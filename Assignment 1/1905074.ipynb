Task e

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
mmscaler = MinMaxScaler()
stdscaler = StandardScaler()
```

```python
def scale(df, MinMax=True):
    if MinMax:
        return mmscaler.fit_transform(df)
    else:
        return stdscaler.fit_transform(df)
```

```python
features.info()
```

```python
target_columns = []
features_mm = features.copy()
features_std = features.copy()
for col in features.columns:
    if features[col].nunique()>2:
        target_columns.append(col)

features_mm[target_columns] = scale(features_mm[target_columns])
features_std[target_columns] = scale(features_std[target_columns],False)
```
